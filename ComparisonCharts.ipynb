```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("FuelConsumption (3).csv")

df.head()

df.info()

df.isnull().sum()

fuel_df = df.groupby("FUELTYPE").mean(numeric_only=True)
fuel_df = fuel_df.reset_index()

fuel_df

plt.figure(figsize=(8,5))
plt.bar(fuel_df["FUELTYPE"], fuel_df["CO2EMISSIONS"])
plt.title("Average CO2 Emissions by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Average CO2 Emissions")
plt.show()

plt.figure(figsize=(8,5))
sns.countplot(x="FUELTYPE", data=df)
plt.title("Number of Vehicles by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Count")
plt.show()

plt.figure(figsize=(12,5))
sns.countplot(x="VEHICLECLASS", data=df)
plt.title("Number of Vehicles by Vehicle Class")
plt.xlabel("Vehicle Class")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()

vehicle_df = df.groupby("VEHICLECLASS").mean(numeric_only=True)
vehicle_df = vehicle_df.reset_index()

vehicle_df

plt.figure(figsize=(12,5))
plt.bar(vehicle_df["VEHICLECLASS"], vehicle_df["CO2EMISSIONS"])
plt.title("Average CO2 Emissions by Vehicle Class")
plt.xlabel("Vehicle Class")
plt.ylabel("Average CO2 Emissions")
plt.xticks(rotation=45)
plt.show()

stacked = vehicle_df[["VEHICLECLASS","FUELCONSUMPTION_CITY","FUELCONSUMPTION_HWY","FUELCONSUMPTION_COMB"]]

stacked.set_index("VEHICLECLASS").plot(kind="bar", stacked=True, figsize=(12,6))

plt.title("Average Fuel Consumption by Vehicle Class")
plt.xlabel("Vehicle Class")
plt.ylabel("Fuel Consumption")
plt.xticks(rotation=45)
plt.legend(["City","Highway","Combined"])
plt.show()
```
