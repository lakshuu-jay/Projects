```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("FuelConsumption.csv")

df.head()

df.info()

df.isnull().sum()

df = df.dropna()

df.select_dtypes(include=np.number).plot(kind='box', figsize=(12,6))
plt.show()

plt.figure(figsize=(8,5))
sns.boxplot(x='FUELTYPE', y='CO2EMISSIONS', data=df)
plt.show()

plt.figure(figsize=(12,5))
sns.boxplot(x='VEHICLECLASS', y='CO2EMISSIONS', data=df)
plt.xticks(rotation=45)
plt.show()

plt.figure(figsize=(8,5))
sns.violinplot(x='FUELTYPE', y='CO2EMISSIONS', data=df)
plt.show()

plt.figure(figsize=(12,5))
sns.violinplot(x='VEHICLECLASS', y='CO2EMISSIONS', data=df)
plt.xticks(rotation=45)
plt.show()

num = df.select_dtypes(include=np.number)

num.plot(kind='density', subplots=True, layout=(2,4), figsize=(12,8))
plt.show()

num.hist(figsize=(12,8))
plt.show()

for i in num.columns:
    plt.figure(figsize=(6,4))
    sns.histplot(df[i], kde=True)
    plt.title(i)
    plt.show()
```
