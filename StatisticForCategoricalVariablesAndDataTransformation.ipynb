```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv('Bestsellers with categories.csv')

data.head()

data.isnull().sum()

data = data.dropna()

print(data['Genre'].value_counts())

print("Median of Genre:", data['Genre'].mode()[0])

data['Genre'].value_counts().plot(kind='bar')
plt.xlabel('Genre')
plt.ylabel('Frequency')
plt.title('Frequency of Genres')
plt.show()

data['Genre'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Distribution of Genres')
plt.ylabel('')
plt.show()

numerical_data = data.select_dtypes(include='number')

sns.boxplot(data=numerical_data)
plt.title('Boxplot of Numerical Features')
plt.show()

normalized_data = (numerical_data - numerical_data.min()) / (numerical_data.max() - numerical_data.min())

print("Normalized Data:")
print(normalized_data.head())

standardized_data = (numerical_data - numerical_data.mean()) / numerical_data.std()

print("Standardized Data:")
print(standardized_data.head())
```
