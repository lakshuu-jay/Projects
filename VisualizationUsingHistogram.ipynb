```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Penguins Data (1).csv")

plt.figure(figsize=(6,4))
plt.hist(df["Culmen Length (mm)"].dropna(), bins=10, color="skyblue")
plt.title("Histogram of Culmen Length")
plt.xlabel("Culmen Length (mm)")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(6,4))
plt.hist(df["Culmen Depth (mm)"].dropna(), bins=8, color="lightgreen")
plt.title("Histogram of Culmen Depth")
plt.xlabel("Culmen Depth (mm)")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(6,4))
plt.hist(df["Flipper Length (mm)"].dropna(), bins=12, color="orange")
plt.title("Histogram of Flipper Length")
plt.xlabel("Flipper Length (mm)")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(6,4))
plt.hist(df["Body Mass (g)"].dropna(), bins=10, color="violet")
plt.title("Histogram of Body Mass")
plt.xlabel("Body Mass (g)")
plt.ylabel("Frequency")
plt.show()
```
