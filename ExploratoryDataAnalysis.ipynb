```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("StudentsPerformance.csv")

print(df.head())
print(df.isnull().sum())

df = df.fillna(df.mode().iloc[0])

print(df.describe())

categorical = ['gender', 'race/ethnicity', 'parental level of education', 'lunch', 'test preparation course']

for col in categorical:
    sns.countplot(data=df, x=col)
    plt.xticks(rotation=45)
    plt.title("Count Plot of " + col)
    plt.show()

df['race/ethnicity'].value_counts().plot.pie(autopct='%1.1f%%')
plt.title("Race/Ethnicity")
plt.ylabel("")
plt.show()

df['parental level of education'].value_counts().plot.pie(autopct='%1.1f%%')
plt.title("Parental Level of Education")
plt.ylabel("")
plt.show()

print(pd.crosstab(df['race/ethnicity'], df['gender']))

print(pd.crosstab(df['lunch'], df['gender']))

numerical = ['math score', 'reading score', 'writing score']

for col in numerical:
    print(col, "Skewness:", df[col].skew())
    sns.histplot(df[col], kde=True)
    plt.title("Distribution of " + col)
    plt.show()

sns.boxplot(data=df, x='gender', y='math score')
plt.title("Gender vs Math Score")
plt.show()

print(df.groupby('gender')['math score'].mean())

correlation = df[numerical].corr()

sns.heatmap(correlation, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
