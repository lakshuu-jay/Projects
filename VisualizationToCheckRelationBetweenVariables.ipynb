```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Penguins Data (2).csv")

df.head()
plt.figure(figsize=(6,4))

plt.scatter(df["Culmen Length (mm)"], df["Body Mass (g)"])

plt.title("Culmen Length vs Body Mass")
plt.xlabel("Culmen Length (mm)")
plt.ylabel("Body Mass (g)")

plt.show()
plt.figure(figsize=(6,4))

plt.scatter(df["Culmen Depth (mm)"], df["Body Mass (g)"])

plt.title("Culmen Depth vs Body Mass")
plt.xlabel("Culmen Depth (mm)")
plt.ylabel("Body Mass (g)")

plt.show()
sns.pairplot(df[[
    "Culmen Length (mm)",
    "Culmen Depth (mm)",
    "Flipper Length (mm)",
    "Body Mass (g)"
]])

plt.show()
area_df = df[["Culmen Length (mm)", "Body Mass (g)"]].dropna()

area_df = area_df.sort_values("Culmen Length (mm)")

plt.figure(figsize=(8,4))

plt.fill_between(
    area_df["Culmen Length (mm)"],
    area_df["Body Mass (g)"]
)

plt.title("Area Graph of Culmen Length and Body Mass")
plt.xlabel("Culmen Length (mm)")
plt.ylabel("Body Mass (g)")

plt.show()
```
