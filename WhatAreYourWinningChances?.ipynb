```python
print("What are your winning chances?")
print("Conditional Probability")

total_balls = 10
white_balls = 3

first_white = white_balls / total_balls
second_white = white_balls / total_balls

print("Probability that the first ball is white:", first_white)
print("Probability that the second ball is white given that the first is white:", second_white)

print("Answer:", second_white)
```
