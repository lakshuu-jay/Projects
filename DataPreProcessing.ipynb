```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

m = pd.read_csv('matches.csv')
d = pd.read_csv('deliveries.csv')

print(m.isnull().sum())
sns.heatmap(m.isnull())
plt.show()
m = m.dropna().drop(columns=['umpire3'], errors='ignore')

print(d.isnull().sum())
sns.heatmap(d.isnull())
plt.show()
d = d.dropna().drop(columns=['player_dismissed', 'dismissal_kind', 'fielder'], errors='ignore')

df = pd.merge(d, m, left_on='match_id', right_on='id')
print(df.head())
```
