```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv('Bestsellers with categories.csv')

data.head()
data.info()
data.isnull().sum()
data = data.dropna()

print("User Rating Variance:", data['User Rating'].var())
print("User Rating Standard Deviation:", data['User Rating'].std())

print("Price Variance:", data['Price'].var())
print("Price Standard Deviation:", data['Price'].std())

plt.hist(data['User Rating'], bins=np.arange(3.0, 5.1, 0.2))
plt.xlabel('User Rating')
plt.ylabel('Frequency')
plt.title('Distribution of User Rating')
plt.show()

plt.hist(data['Price'], bins=np.arange(0, 110, 10))
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.title('Distribution of Price')
plt.show()
```
