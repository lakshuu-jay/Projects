```python
import math

lam = 10

def poisson_probability(k):
    return math.exp(-lam) * lam**k / math.factorial(k)

p_12_or_more = 1 - sum(poisson_probability(k) for k in range(12))

p_12_to_18 = sum(poisson_probability(k) for k in range(12, 19))

print("Probability of 12 or more rainy days:", p_12_or_more)
print("Percentage:", p_12_or_more * 100, "%")

print("Probability of 12 to 18 rainy days:", p_12_to_18)
print("Percentage:", p_12_to_18 * 100, "%")
```
