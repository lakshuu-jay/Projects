```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv('Bestsellers with categories.csv')

data.head()

data.info()

data.isnull().sum()

data = data.dropna()

print("10% Quantile of Price:", data['Price'].quantile(0.10))

Q1 = data['Price'].quantile(0.25)
Q2 = data['Price'].quantile(0.50)
Q3 = data['Price'].quantile(0.75)
IQR = Q3 - Q1

print("Price Q1:", Q1)
print("Price Q2:", Q2)
print("Price Q3:", Q3)
print("Price IQR:", IQR)

Q1 = data['User Rating'].quantile(0.25)
Q2 = data['User Rating'].quantile(0.50)
Q3 = data['User Rating'].quantile(0.75)
IQR = Q3 - Q1

print("User Rating Q1:", Q1)
print("User Rating Q2:", Q2)
print("User Rating Q3:", Q3)
print("User Rating IQR:", IQR)

sns.boxplot(x=data['Price'])
plt.title('Boxplot of Price')
plt.show()
sns.boxplot(x=data['User Rating'])
plt.title('Boxplot of User Rating')
plt.show()
```
