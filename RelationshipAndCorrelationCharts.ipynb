```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv('FuelConsumption.csv')

data.head()
data.info()

data.isnull().any()

data = data.dropna()

sns.scatterplot(data=data, x='FUELCONSUMPTION_COMB_MPG', y='CO2EMISSIONS')
plt.show()
sns.scatterplot(data=data, x='FUELCONSUMPTION_COMB_MPG', y='CO2EMISSIONS', hue='FUELTYPE')
plt.show()

fig, ax = plt.subplots(figsize=(8,8))
sns.scatterplot(data=data,
                x='FUELCONSUMPTION_COMB_MPG',
                y='CO2EMISSIONS',
                hue='FUELTYPE',
                size='ENGINESIZE',
                sizes=(20,300),
                palette='bright')
plt.show()

sns.heatmap(data.corr(numeric_only=True), annot=True)
plt.show()

sns.relplot(data=data,
            x='FUELCONSUMPTION_COMB_MPG',
            y='CO2EMISSIONS',
            col='FUELTYPE',
            col_wrap=3,
            height=4)
plt.show()

sns.pairplot(data, hue='FUELTYPE')
plt.show()
```
