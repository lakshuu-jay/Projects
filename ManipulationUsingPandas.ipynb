```python
import pandas as pd
data = {
    "Name": ["Roy", "Anna", "John", "Sara"],
    "Department": ["IT", "HR", "Finance", "Marketing"],
    "Salary": [5000, 4500, 6000, 5200]
}

df = pd.DataFrame(data)
print(df)
print(df[df["Salary"] > 5000])
df["Bonus"] = df["Salary"] * 0.10

print(df)
```
