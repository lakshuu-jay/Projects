```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

july = pd.DataFrame({
    "Day": ["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"],
    "New Births": [12,15,11,9,1,9,21]
})

plt.figure(figsize=(7,5))
plt.bar(july["Day"], july["New Births"], color="skyblue", width=0.6, edgecolor="black")
plt.title("Birth Rate - July")
plt.xlabel("Day")
plt.ylabel("New Births")
plt.show()

august = pd.DataFrame({
    "Day": ["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"],
    "New Births": [17,5,2,11,1,8,29]
})

plt.figure(figsize=(7,5))
plt.bar(august["Day"], august["New Births"], color="orange", width=0.6, edgecolor="black")
plt.title("Birth Rate - August")
plt.xlabel("Day")
plt.ylabel("New Births")
plt.show()

x = np.arange(len(july["Day"]))
width = 0.35

plt.figure(figsize=(8,5))
plt.bar(x - width/2, july["New Births"], width, color="skyblue", edgecolor="black", label="July")
plt.bar(x + width/2, august["New Births"], width, color="orange", edgecolor="black", label="August")

plt.xticks(x, july["Day"])
plt.xlabel("Day")
plt.ylabel("New Births")
plt.title("Birth Rate Comparison: July vs August")
plt.legend()
plt.show()
```
