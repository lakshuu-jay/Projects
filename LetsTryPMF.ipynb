```python
import math

n = 10
p = 0.5

probability = 0

for k in range(2, 5):
    pmf = math.comb(n, k) * (p ** k) * ((1 - p) ** (n - k))
    probability += pmf
    print(f"P(X = {k}) = {pmf:.4f}")

print(f"\nProbability of getting between 2 and 4 heads = {probability:.4f}")
print(f"Percentage = {probability * 100:.2f}%")
```
