```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('FuelConsumption (1).csv')

data.head()

data.info()

print(data.isnull().sum())

data = data.fillna(0)

fuel_data = data.groupby('FUELTYPE').mean(numeric_only=True)

fuel_data = fuel_data.reset_index()

print(fuel_data)

plt.figure(figsize=(6,4))

plt.bar(fuel_data['FUELTYPE'], fuel_data['CO2EMISSIONS'])

plt.title("Average CO2 Emissions by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Average CO2 Emissions")

for i in range(len(fuel_data)):
    plt.text(i,
             fuel_data['CO2EMISSIONS'][i],
             round(fuel_data['CO2EMISSIONS'][i],1),
             ha='center')

plt.show()
```
