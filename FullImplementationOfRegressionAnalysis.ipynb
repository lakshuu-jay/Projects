```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('xydataset.csv')

x = data.iloc[1:,1].astype(float).values
y = data.iloc[1:,2].astype(float).values

m = 0
c = 0
lr = 0.0001

for i in range(1000):
    yp = m*x + c
    dm = (-2/len(x))*sum(x*(y-yp))
    dc = (-2/len(x))*sum(y-yp)
    m = m - lr*dm
    c = c - lr*dc

yp = m*x + c

print("y =", m, "x +", c)

plt.scatter(x,y)
plt.plot(x,yp)
plt.show()
```
