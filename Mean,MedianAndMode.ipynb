```python
import pandas as pd
import numpy as np

data = pd.read_csv('Bestsellers with categories.csv')

data.head()

data.info()

data.isnull().sum()

data = data.fillna(0)

categorical = ['Name', 'Author', 'Genre']
quantitative = ['User Rating', 'Reviews', 'Price', 'Year']

print("Categorical Features:")
print(categorical)

print("Quantitative Features:")
print(quantitative)

print("Mean User Rating:", data['User Rating'].mean())
print("Median User Rating:", data['User Rating'].median())
print("Mode User Rating:", data['User Rating'].mode()[0])

print("Mean Price:", data['Price'].mean())
print("Median Price:", data['Price'].median())
print("Mode Price:", data['Price'].mode()[0])

print("Mean Reviews:", data['Reviews'].mean())
print("Median Reviews:", data['Reviews'].median())
print("Mode Reviews:", data['Reviews'].mode()[0])
```
