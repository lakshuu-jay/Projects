```python
import numpy as np
arr = np.arange(10)
print("Original array:", arr)
arr2 = arr.copy()
arr2[arr2 % 2 == 1] = -1
print("Array with odd numbers replaced:", arr2)
arr3 = arr.reshape(2, 5)
print("2D array with two rows:\n", arr3)
even_sum = np.sum(arr[arr % 2 == 0])
print("Sum of even numbers:", even_sum)
```
