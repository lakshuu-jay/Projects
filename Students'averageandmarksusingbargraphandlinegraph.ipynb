```python
import matplotlib.pyplot as plt

student_names = ["Alex","Martin","Jade","Bianca","James","Alice"]
student_marks = [23,45,30,50,48,38]
student_perc = []

for i in student_marks:
  result = (i/50)*100
  student_perc.append(result)

print(student_perc)

def marks_linechart():
  plt.plot(student_names,student_marks)
  plt.title("Students Annual Results")
  plt.xlabel("Names")
  plt.ylabel("Marks")
  plt.show()

marks_linechart()

def bar_chart():
  plt.bar(student_names,student_perc)
  plt.title("Students Annual Results")
  plt.xlabel("Names")
  plt.ylabel("Percentage")
  plt.show()

bar_chart()
```
