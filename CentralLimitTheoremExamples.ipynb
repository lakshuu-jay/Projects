```python
import random
import statistics

heights = [150, 160, 170, 180]
print("Height:", statistics.mean(random.sample(heights, 2)))

dice = [random.randint(1, 6) for i in range(5)]
print("Dice:", statistics.mean(dice))

scores = [50, 60, 70, 80, 90]
print("Scores:", statistics.mean(random.sample(scores, 3)))

temp = [25, 27, 29, 31, 33]
print("Temperature:", statistics.mean(random.sample(temp, 3)))

money = [10, 20, 30, 40, 50]
print("Money:", statistics.mean(random.sample(money, 3)))
```
