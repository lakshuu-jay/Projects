```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("FuelConsumption (4).csv")

df.head()

df.info()

df.isnull().sum()

df = df.dropna()

fuel = df["FUELTYPE"].value_counts()

plt.figure(figsize=(7,7))
plt.pie(fuel,
        labels=fuel.index,
        autopct="%1.1f%%")
plt.title("Fuel Type")
plt.show()

plt.figure(figsize=(7,7))
plt.pie(fuel,
        labels=fuel.index,
        autopct="%1.1f%%",
        wedgeprops={"edgecolor":"black","linewidth":2})
plt.title("Fuel Type (Customized)")
plt.show()

vehicle = df["VEHICLECLASS"].value_counts()

plt.figure(figsize=(8,8))
plt.pie(vehicle,
        labels=vehicle.index,
        autopct="%1.1f%%")
plt.title("Vehicle Class")
plt.show()

plt.figure(figsize=(12,5))
plt.plot(df["CO2EMISSIONS"])
plt.title("CO2 Emissions")
plt.xlabel("Vehicles")
plt.ylabel("CO2 Emissions")
plt.show()
```
