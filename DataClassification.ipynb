```python
import pandas as pd

data = pd.read_csv("glass.csv")

print(data.head())

target = data.iloc[:, -1]

print("Classes:", target.unique())
print("Number of Classes:", target.nunique())

if target.nunique() == 2:
    print("This is a Binary Classification dataset.")
else:
    print("This is a Multi-Class Classification dataset.")

print("The dataset contains", target.nunique(), "different classes, so it is a Multi-Class Classification dataset.")
```
